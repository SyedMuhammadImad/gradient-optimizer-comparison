# optimizer comparison custom optimizer comparison

Four manually implemented optimizers train the same 25-feature MLP from the same initialization and batch order: SGD, momentum SGD, RMSProp and Adam. Parameter trajectories match independent PyTorch references, including weight decay and intermittent missing gradients. Adam uses per-parameter step counts; RMSProp adds epsilon outside the square root.


```python
import json,sys
from pathlib import Path
ROOT=next(p for p in [Path.cwd(),*Path.cwd().parents] if (p/'project.json').exists())
sys.path.insert(0,str(ROOT))
import numpy as np
import models
```

## Implementation check

This mathematical smoke check is separate from measured classifier accuracy.


```python
import torch
from experiment_core import as_tensor,torch_factories,SPECS
kind='optimizer'
factory=next(iter(torch_factories(kind).values()))[0]
model=factory().eval()
with torch.no_grad():
    probs=model(as_tensor(np.zeros((3,SPECS[kind][0])),kind)).softmax(dim=1)
assert torch.isfinite(probs).all() and torch.allclose(probs.sum(dim=1),torch.ones(3))
```
